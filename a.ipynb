```python
for c in ['quantity', 'unit_price_usd', 'total_price_usd']:
    q1, q3 = df[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    n = ((df[c] < q1 - 1.5 * iqr) | (df[c] > q3 + 1.5 * iqr)).sum()
    print(c, n)
```

```python

print("=== TOP 10 unit_price_usd ===")
print(df.nlargest(10, 'unit_price_usd')[['order_id', 'unit_price_usd', 'total_price_usd', 'quantity']])

print("\n=== TOP 10 total_price_usd ===")
print(df.nlargest(10, 'total_price_usd')[['order_id', 'unit_price_usd', 'total_price_usd', 'quantity']])
print("\n=== TOP 10 quantity ===")
print(df.nlargest(10, 'quantity')[['order_id', 'unit_price_usd', 'total_price_usd', 'quantity']])
```

```python
df['subtotal'] = df['unit_price_usd'] * df['quantity']
df['expected_total'] = (df['subtotal'] - df['discount_amount_usd']).round(2)

# Difference and ratio
df['difference'] = (df['total_price_usd'] - df['expected_total']).round(2)
df['ratio'] = (df['total_price_usd'] / df['expected_total'].replace(0, np.nan)).round(2)

# Mismatches: total is more than 1.5× the expected
mismatches = df[df['ratio'] > 1.5]

print(f"Mismatched rows: {len(mismatches)}")
print(mismatches[['order_id', 'unit_price_usd', 'quantity', 'discount_amount_usd',
                  'subtotal', 'expected_total', 'total_price_usd',
                  'difference', 'ratio']].head(30))
```

```python
df['order_date'].dtype 
```

```python

print("Before:", df['order_date'].dtype)
df['order_date'] = pd.to_datetime(
    df['order_date'],
    errors='coerce',
    dayfirst=True,
    format='mixed'
)
bad = df['order_date'].isnull().sum()
df = df.dropna(subset=['order_date']).reset_index(drop=True)
df['order_year']  = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_day']   = df['order_date'].dt.day

print("After:", df['order_date'].dtype)
print("NaT dropped:", bad)
```

```python

print("Missing city before:", df['city'].isnull().sum())
city_mode = df['city'].mode()[0]
df['city'] = df['city'].fillna(city_mode)
print(f"Mode used: '{city_mode}'")
print("Missing city after:", df['city'].isnull().sum())
```

```python

subtotal = df['unit_price_usd'] * df['quantity']
expected = (subtotal - df['discount_amount_usd']).round(2)
ratio    = df['total_price_usd'] / expected.replace(0, np.nan)
mismatch_mask = ratio > 1.5
print("Mismatched rows detected:", mismatch_mask.sum())
df = df[~mismatch_mask].reset_index(drop=True)
print("Rows after removal:", len(df))
```

```python
df.to_csv("cleanedecommercedataset.csv", index=False)
```
